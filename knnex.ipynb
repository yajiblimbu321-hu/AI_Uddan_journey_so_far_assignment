```python
x=df[[
    'Attendance', 'PreviousMarks', 'AssignmentCompletion',
       'QuizScore'
]]
```

```python
y=df[
    'Pass'
]
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler   
from sklearn.neighbors import KNeighborsClassifier   
```

```python
x_tro,x_teo,y_tro,y_teo=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
scaler=StandardScaler()
x_tr=scaler.fit_transform(x_tro)
x_te=scaler.transform(x_teo)
```

```python
x_tr
```

```python
model=KNeighborsClassifier(
    n_neighbors=3
)
model.fit(x_tr,y_tro)
```

```python
prediction=model.predict(x_te)
prediction
```

```python
data1=pd.DataFrame({
    "Acyual":y_teo,
    "Predicted":prediction
})
data1
```

```python
model.predict_proba(x_te)
```

```python
from sklearn.metrics import accuracy_score,precision_score,f1_score,recall_score
```

```python
print("Accuracy",accuracy_score(y_teo,prediction))
print("Precision",precision_score(y_teo,prediction))
print("F1 score",f1_score(y_teo,prediction))
print("Recall",recall_score(y_teo,prediction))
```

```python
distance, index = model.kneighbors(new_scaled)
# kneighbors() returns TWO things: how far each neighbor is (distance),
# and WHERE each neighbor sits in the training data (index/position).
# We need distance to show/verify "closeness", and index to trace back
# to WHO that neighbor actually is (since KNN only stores numbers, not names).

print(index)
# index gives POSITIONS within the training set (not the original df row numbers),
# because kneighbors() has no idea about df's original row labels — it only
# knows "1st row I was trained on, 2nd row, 3rd row..." Since train_test_split
# shuffled the data, position ≠ original df row number anymore.

datatr = pd.DataFrame(x_tro, columns=[...])
# Rebuild the training data as a DataFrame (not numpy) so it keeps its
# ORIGINAL df index attached — this becomes our "translator" between
# position and original identity.

actual_row_numbers = datatr.index[index[0]]
# index[0] = positions from kneighbors (e.g. [1,14,9])
# datatr.index[...] converts each POSITION into its ORIGINAL df row label
# — without this step, we'd have no way to know which real student
# each neighbor actually is.

df.loc[actual_row_numbers]
# Finally, use those real original row numbers to pull the actual student
# rows (with names) from the ORIGINAL df — this answers the real question:
# "which existing students are most similar to this new student?"
```
