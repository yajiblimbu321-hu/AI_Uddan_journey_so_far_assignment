```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('datasets/Question2.csv')
df
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df.columns
```

```python
x=df[[ 'Attendance', 'MathScore', 'EnglishScore', 'QuizScore',
       'MissedClasses']]#i think quiz and missed class also required 
x=x.to_numpy()
```

```python
y=df[['MathSupport', 'EnglishSupport', 'AttendanceSupport']]
y=y.to_numpy()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_tr,x_te,y_tr,y_te=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
x_tr
```

```python
y_tr
y_te.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.multiclass import OneVsRestClassifier
```

```python
model=OneVsRestClassifier(LogisticRegression())
```

```python
model.fit(x_tr,y_tr)
```

```python
prediction=model.predict(x_te)
```

```python
model.predict_proba(x_te)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_te, prediction)
```

```python
for i, estimator in enumerate(model.estimators_):
    
    print(estimator.coef_)
    print()
```

```python
actual_df = pd.DataFrame(y_te, columns=['Label1', 'Label2', 'Label3'])
predicted_df = pd.DataFrame(prediction, columns=['Label1', 'Label2', 'Label3'])

print("Actual:")
print(actual_df)
print("Predicted:")
print(predicted_df)
```

```python
new_std=pd.DataFrame({
    'Attendance':[55],
    'MathScore':[66],
    'EnglishScore':[56],
    'QuizScore':[67],
    'MissedClasses':[4]
})
new_std=new_std.to_numpy()
new_pre=model.predict(new_std)
data2=pd.DataFrame(new_pre,columns=['MathSupport', 'EnglishSupport', 'AttendanceSupport'])
data2
```
