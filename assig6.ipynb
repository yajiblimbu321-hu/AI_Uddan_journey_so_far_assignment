```python
data['PreviousMarks']=data['PreviousMarks'].fillna(data['PreviousMarks'].quantile(0.5))
```

```python
data.isna().sum()
```

```python
data.info()
```

```python
numericd=['StudyHours',"Attendance",'PreviousMarks','FinalMarks']
```

```python
for c in numericd:
    data[c]=pd.to_numeric(data[c],errors="coerce")
    data[c]=data[c].fillna(data[c].quantile(0.5))
```

```python
data
```

```python
X=data[['StudyHours', 'Attendance', 'PreviousMarks']]
X=X.to_numpy()
X.shape
```

```python
Y=data["FinalMarks"]
Y.to_numpy()
Y.shape
```

```python
from sklearn.model_selection import train_test_split
f_train,f_test,l_train,_test=train_test_split(X,Y,test_size=0.2, random_state=42)
```

```python
model=LinearRegression()
model.fit(f_train,l_train)
```

```python
print("bias or y when x=0 is",model.intercept_)
```

```python
print("Coefficient of StudyHours:", model.coef_[0])
print("Coefficient of Attendance:", model.coef_[1])
print("Coefficient of PreviousMarks:", model.coef_[2])
```

```python
arr=np.array([
    [5,90,70],
    [7,95,82],
    [3,75,55]
])
arr
```

```python
predictions=model.predict(arr)
print(predictions)
```

```python
students = ['1', '2', '3']
for s, p in zip(students, predictions):
    print(f"Student {s} =>Predicted Final Marks: {p:.2f}")
```
