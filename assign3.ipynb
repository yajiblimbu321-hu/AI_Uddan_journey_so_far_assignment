```python
d.columns
```

```python
outliers_c=['course_fee','attendance_percent','assignment_score', 'final_score']
```

```python
dataa={}
for columns in outliers_c:
    q1=d[columns].quantile(0.25)
    print(f"q1 for column:{columns}={q1}")
    q3=d[columns].quantile(0.75)
    print(f"q3 for column:{columns}={q3}")
    median=d[columns].quantile(0.5)
    iqr=q3-q1
    print(f"iqr for column:{columns}={iqr}")
    lower=q1-1.5*iqr
    print(f"lower_bound for column:{columns}={lower}")
    upper=q3+1.5*iqr
    print(f"upper_bound for column:{columns}={upper}")
    print()
    dataa[columns]={
        "q1":q1,
        "q3":q3,
        "iqr":iqr,
        "lower_bound":lower,
        "upper_bound":upper,
        "median":median
        
    }
    
    
```

```python
print(dataa)
print()
print(dataa['course_fee'])
print()

print(dataa['course_fee']['q1'])
```

```python
for columns in outliers_c:
    print(f'removing outlier from {columns} using drop')
    d=d.drop(d.loc[(d[columns]<dataa[columns]['lower_bound'])|(d[columns]>dataa[columns]['upper_bound'])].index)
    print(d[columns])
    print("----"*100)
```

```python
import numpy as np
```

```python
for columns in outliers_c:
    print(f'removing outlier from {columns} using meidan')
    d.loc[(d[columns]<dataa[columns]['lower_bound'])|(d[columns]>dataa[columns]['upper_bound'])]=np.nan
    d[columns]=d[columns].fillna(dataa[columns]['median'])
    print(d[columns])
    print("----"*100)
```

```python
for columns in outliers_c:
    print(f"Clipping outliers in {columns}")

    d[columns] = d[columns].clip(lower=dataa[columns]["lower_bound"],upper=dataa[columns]["upper_bound"]
    )

    print(d[columns])
    print("----" * 100)
```

```python
for columns in outliers_c:
    print(f"Using Z-score for column: {columns}")

    mean = d[columns].mean()
    std = d[columns].std()
    a="zscore-"+columns

    d[a] = (d[columns] - mean) / std

    print("Mean:", mean)
    print("Standard Deviation:", std)

    outliers = d.loc[
        (d[a] > 3) |
        (d[a] < -3)
    ]

    print(outliers)
    print("-" * 50)
```
